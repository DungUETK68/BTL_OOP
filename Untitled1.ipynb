```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix
)
import joblib
import xgboost as xgb
import matplotlib.pyplot as plt
import seaborn as sns

# ---------- CONFIG ----------
TRAIN_PATH = "/content/train_mini.csv"
VAL_PATH   = "/content/val_mini.xlsx" # Changed to .xlsx based on uploaded file
OUTPUT_MODEL = "/content/xgb_multiclass_pipeline.joblib" # Changed path to /content/
RANDOM_STATE = 42
EARLY_STOPPING_ROUNDS = 50
NUM_CLASSES = 10
# ---------------------------

def load_data(path):
    if not os.path.exists(path):
        raise FileNotFoundError(f"File not found: {path}")
    if path.endswith('.csv'):
        df = pd.read_csv(path)
    elif path.endswith('.xlsx'):
        df = pd.read_excel(path)
    else:
        raise ValueError("Unsupported file format. Please provide a .csv or .xlsx file.")
    return df

def build_preprocessor(X):
    numeric_cols = X.select_dtypes(include=['int64','float64']).columns.tolist()
    cat_cols = X.select_dtypes(include=['object','category','bool']).columns.tolist()

    low_cardinality = [c for c in cat_cols if X[c].nunique() <= 20]
    high_cardinality = [c for c in cat_cols if X[c].nunique() > 20]

    numeric_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='mean')),
        ('scaler', StandardScaler())
    ])

    low_cat_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='constant', fill_value='__NA__')),
        ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    ])

    high_cat_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='constant', fill_value='__NA__')),
        ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
    ])

    transformers = []
    if numeric_cols:
        transformers.append(('num', numeric_transformer, numeric_cols))
    if low_cardinality:
        transformers.append(('lowcat', low_cat_transformer, low_cardinality))
    if high_cardinality:
        transformers.append(('highcat', high_cat_transformer, high_cardinality))

    return ColumnTransformer(transformers, remainder='drop')

def plot_confusion_matrix(cm, classes, outpath=None):
    plt.figure(figsize=(8,6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=classes, yticklabels=classes)
    plt.ylabel('True label')
    plt.xlabel('Predicted label')
    plt.title('Confusion Matrix')
    plt.tight_layout()
    if outpath:
        plt.savefig(outpath)
    plt.show()

def main():
    print("Loading training & validation data...")
    train_df = load_data(TRAIN_PATH)
    val_df = load_data(VAL_PATH)

    print("Train shape:", train_df.shape)
    print("Val shape:", val_df.shape)

    # ------------------ TRAIN ------------------
    if 'class' not in train_df.columns:
        raise KeyError("Không tìm thấy cột 'class' trong train file.")

    X_train = train_df.drop(columns=['class'])
    y_train = train_df['class'].astype(int)

    # ------------------ VALIDATION ------------------
    if 'class' not in val_df.columns:
        raise KeyError("Không tìm thấy cột 'class' trong val file.")

    X_val = val_df.drop(columns=['class'])
    y_val = val_df['class'].astype(int)

    unique_labels = sorted(y_train.unique())
    print("Unique labels in train:", unique_labels)

    # build preprocessor
    preprocessor = build_preprocessor(X_train)
    preprocessor.fit(X_train)

    X_train_trans = preprocessor.transform(X_train)
    X_val_trans = preprocessor.transform(X_val)

    # ------------------ XGBoost model ------------------
    xgb_clf = xgb.XGBClassifier(
        objective='multi:softprob',
        num_class=NUM_CLASSES,
        n_estimators=1000,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric='mlogloss',
        random_state=RANDOM_STATE,
        n_jobs=-1
    )

    print("Training XGBoost with early stopping...")
    xgb_clf.fit(
        X_train_trans, y_train,
        eval_set=[(X_val_trans, y_val)],
        early_stopping_rounds=EARLY_STOPPING_ROUNDS,
        verbose=100
    )

    # ------------------ Evaluation ------------------
    print("\nEvaluating...")
    y_pred = xgb_clf.predict(X_val_trans)

    print("Accuracy:", accuracy_score(y_val, y_pred))
    print("Weighted F1:", f1_score(y_val, y_pred, average='weighted'))
    print("\nClassification report:")
    print(classification_report(y_val, y_pred))

    cm = confusion_matrix(y_val, y_pred)
    plot_confusion_matrix(
        cm,
        classes=[str(c) for c in unique_labels],
        outpath="/content/confusion_matrix.png" # Changed path from /mnt/data/ to /content/
    )

    print("Saved confusion matrix \u2192 /content/confusion_matrix.png")

    # ------------------ Save full pipeline ------------------
    full_pipeline = Pipeline([
        ('preproc', preprocessor),
        ('xgb', xgb_clf)
    ])

    joblib.dump(full_pipeline, OUTPUT_MODEL)
    print(f"Saved model \u2192 {OUTPUT_MODEL}")

if __name__ == "__main__":
    main()
```
